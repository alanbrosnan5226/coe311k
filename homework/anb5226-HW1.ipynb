```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
def MonteCarloApproximation(numsamples):
    hits= 0
    for i in range (numsamples):
        x= random.uniform(-1,1)
        y= random.uniform(-1,1)
        
        if (x*x + y*y <1):
                hits+=1
        
    pi_approx= (4*hits)/numsamples
    return pi_approx
```

```python
MonteCarloList = []
Samples=[]
pi_actual= round(np.pi, 15) 
for numsamples in range (10,10001,20):
    pi_est= MonteCarloApproximation(numsamples)
    Error= abs(pi_est - pi_actual)
    MonteCarloList.append (Error)
    Samples.append(numsamples)
```

```python
plt.plot (Samples, MonteCarloList)
plt.xlabel ("Number of Samples")
plt.ylabel ("Error Size")
plt.title ("Plotting Monte Carlo Pi")
plt.show()
```
