```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def SinApprox (x, samples):
    x= np.asarray(x,dtype=float) 
    
    approx_value=0
    for n in range(samples):
        approx_value+=((-1)**n)*((x**(2*n+1))/(math.factorial(2*n+1)))
    return approx_value
```

```python
x_values= np.arange(0, 10.1, 0.1)
real_sin= np.sin(x_values)

approx_sinlist={}
error={}

for n in range (1,5):
    approx_sinlist[n] = SinApprox(x_values,n)
    error[n]= np.abs(approx_sinlist[n]-real_sin)
```

```python
plt.figure()
for n in range (1,5):
    plt.plot(x_values, error[n], label=f"{n} terms(s)")
plt.xlabel("x")
plt.ylabel("Absolute Error")
plt.yscale("log")
plt.title("Truncation Error for Taylor Series sin(x) approximation")
plt.legend()
plt.grid(True)
plt.show()
```

Discussion: As more terms are added to the approximation typically the more accurate it is. For example at lower numbers the 4 term is the most accurate, however at a certain point the more values are at the more inaccurate the higher terms become. The error does not decrease uniformally, for example the error is more accurate when smaller values are being used and they tend to get slowly more inaccurate with larger values. Approximation tends to be accurate near values of 0 and requires more terms to maintain accuracy as x increases.
