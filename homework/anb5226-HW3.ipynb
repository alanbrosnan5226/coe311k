```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def SinApprox (x, samples):
    x= np.asarray(x,dtype=float) 
    
    approx_value=0
    for n in range(samples):
        approx_value+=((-1)**n)*((x**(2*n+1))/(math.factorial(2*n+1)))
    return approx_value
```

```python
x_values= np.arange(0, 10.1, .1)
real_sin= np.sin(x_values)

approx_sinlist={}
error={}

for n in range (1,5):
    approx_sinlist[n] = SinApprox(x_values,n)
    error[n]= np.abs(approx_sinlist[n]-real_sin)
```

```python
plt.figure()
for n in range (1,5):
    plt.plot(x_values, error[n], label=f"{n} terms(s)")
plt.xlabel("x")
plt.ylabel("Absolute Error")

plt.title("Truncation Error for Taylor Series sin(x) approximation")
plt.legend()
plt.grid(True)
plt.show()
```

Discussion: As more terms are added to the approximation typically the more accurate it is. For example at lower numbers the 4 term is the most accurate, however at a certain point the greater values are less accurate because higher order terms are ignored. The error does not decrease uniformaly, for example the error smaller when smaller x-values are being used and they tend to get slowly more inaccurate with larger values, then there's an exponential increase in their inaccuracy when values get larger. Approximation tends to be accurate near values of 0 and requires more terms to maintain accuracy as x increases.
