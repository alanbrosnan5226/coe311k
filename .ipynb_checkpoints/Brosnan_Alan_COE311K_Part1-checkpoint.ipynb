# Section 1: Introduction & Model Selection

Option A:

Spring-Mass System: mf''(t) + cf'(t) + k*f(t) = 0

My Application:
m(d^2*y/dt^2) + c(dy/dt) + ky = F.


```python
#Conversion to first ODE
import numpy as np
import matplotlib.pyplot as plt

# Constant Parameters from research
m = 42000       # Mass (kg)
c = 10800       # Damping (N*s/m)
k = 16800       # Spring Constant (N/m)
f = 71171.5    # Maneuver Force (N)

# Converts the second ODE into a system of first ODE
def df(t, x, m, c, k, f):
    x1, x2 = x
    dx1 = x2
    dx2 = (f/m) - (c/m)*x2 - (k/m)*x1
    return np.array([dx1, dx2])

print("FIRST ODE SYSTEM")
print("dx1 = x2")
print("dx2 = (f/m)-(c/m)*x2-(k/m)*x1")

x_zero = np.array([0.0, 0.0])  # y=0, v=0 at t=0
print("First ODE System results at t=0:")
print(df(0, x_zero, m, c, k, f))
```

Physical Meaning of Variables:

In this equation the m is characteristic of the participating mass (slosh mass), the fluid that is being “sloshed”. C is representative of the dampening coefficient which is the boundary layer and the baffles. Essentially, this means that c is representative of the energy that is lost as viscous friction occurs when the propellant is sloshed up against the interior of the fuel storage container. Since liquid is naturally low dampening engineers have added slosh baffles, rings inside the tank, to artificially increase the energy loss or the c constant. Finally, the k constant which acts as the restoring force or spring constant, serves as the gravity in our case. Since there is no traditional spring in fluid containers of rockets, the “restoring force” in our case is the acceleration caused by gravity. As the fluid sloshes, gravity pulls the fluid back to it’s equilibrium state. The restoring force is directly correlated to the G-force of thrust, essentially the more the rocket accelerates upward the stronger the G-force pulls and the stiffer the liquid becomes thus increasing the k constant. The F(t) is the manuever force that is being applied laterally to the tank of liquid fuel. As the engine gimbles it will apply lateral movement which will cause the liquid in the tank to slosh.

# Section 3: Numerical Methods Implementation

```python
#Euler's Forward Method
#creating the for loop
#at least three different step sizes (e.g., h = 0.1, 0.05, 0.01)
#EXPLAIN THE TIME INTERVAL
x_end = 40    # End of the interval
def euler_forward(m, c, k, f, y0, v0, t_start, t_end, h):
    t_values = np.arange(t_start, t_end + h, h)
    y_values = np.zeros(len(t_values))
    v_values = np.zeros(len(t_values))
    
    y_values[0] = y0
    v_values[0] = v0
    
    for i in range(1, len(t_values)):
        x = np.array([y_values[i-1], v_values[i-1]])
        dx = df(t_values[i-1], x, m, c, k, f)
        
        y_values[i] = y_values[i-1] + h * dx[0]
        v_values[i] = v_values[i-1] + h * dx[1]
    
    return t_values, y_values, v_values
```

EXPLAIN THE EULERS:
After splitting up the 2nd ODE for spring mass oscillation into a system of two first order differential equations, Eulers method is applied to determine the next state of the function. Eulers work's by using the slope at the current time to approximate the next point. The function can take both the position and velocity equations for the system of first order differential equations. In my application it therefore calculates the position of the slosh mass (liquid oxgyen propellent) as it moves throughout the tank and the velocity of its oscillations. 

```python
#Fourth Order Range Kutta
# Define the ODE function, for example, v_n + h * (f-c*v_n-k*y_n)/m
# Set initial conditions and parameters
t0 = 0.0  # Initial time
tf = 40.0  # Final time

def runge_kutta_4(m, c, k, f, y0, v0, t0, tf, h):
    t = np.arange(t0, tf + h, h)
    x = np.zeros((len(t), 2))  # [y, v]
    
    # Initial conditions
    x[0] = [y0, v0]
    
    for i in range(len(t) - 1):
        k1 = df(t[i], x[i], m, c, k, f)
        k2 = df(t[i] + h/2, x[i] + h/2 * k1, m, c, k, f)
        k3 = df(t[i] + h/2, x[i] + h/2 * k2, m, c, k, f)
        k4 = df(t[i] + h, x[i] + h * k3, m, c, k, f)
        
        x[i+1] = x[i] + h * (k1 + 2*k2 + 2*k3 + k4) / 6
    
    return t, x[:,0], x[:,1]  # return y and v separately
```

EXPLAIN THE RK4: 
Runge Kutta was applied to provide a new layer of accuracy and stability over Eulers method. RK4 is more accurate due to the fact it calculates the slope at 4 different points (beginning, midpoint using k1, midpoint using k2, and the end of the interval) to calculate the slope of the next point. In my application it works the same as Eulers to calculate the velocity and position of the sloshing mass as it oscillates, but it provides a new layer of accuracy that Euler's forward cannot.


# Section 4: Solutions & Comparison

```python
#EULERS PLOTTING for positiion
plt.figure()
h_steps = [0.1, 0.05, 0.01] #defined here for all the cells after this
for h in h_steps:
    t, y, v = euler_forward(m, c, k, f, y0, v0, x_start, x_end, h)
    plt.plot(t, y, label=f"h = {h}")
# Plot the results
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
y_eq = f / k #position equilibrium
plt.axhline(y_eq, linestyle='--', label="Position Equilibrium")
plt.title('Damped Spring Mass System Eulers Forward (POSITION)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Eulers Plotting for Velcocity
plt.figure()

for h in h_steps:
    t, y, v = euler_forward(m, c, k, f, y0, v0, x_start, x_end, h)
    plt.plot(t, v, label=f"h = {h}")
# Plot the results
plt.xlabel('Time (s)')
plt.ylabel('Velocity (m/s)')
v_eq = 0 #equilibrium for velocity (obviously its zero idk why it took me so long to think about it)
plt.axhline(v_eq, linestyle='--', label="Velocity Equilibrium (v=0)")
plt.title('Damped Spring Mass System Eulers Forward (VELOCITY)')
plt.legend()
plt.grid(True)
plt.show()
```

#TALK ABOUT THE EULERS GRAPH AND THE ACCURACY IN IT
Eulers graph can be used to model both the position and the accuracy of the ODE system. By calculating the slope at the next point in using a previous point we can create the model that is demonstrated. The models it has created are relatively accurate, demonstrated by the oscillating trend as velocity falls back to 0 and the position oscillates to the equilibrium. The accuracy of the eulers forward is extremely dependent on the h step size. When playing with higher h values, the graph comes out jagged and crooked, failing to capture the smooth oscillation hills that are characterisitc of liquid oxidizer slosh. However, the smaller the step size, the more accurate the model is, which can be demonstrated as the smallest h step has a more accurate graph oscillating closer to the equilibrium. The smaller the step size the more computational time the graph will take but the more accurate it will be, demonstrated by the smoothness. 

```python
#RK4 PLotting (POSITION):
plt.figure()
for h in h_steps:
    #Plot one at a time 
    t_rk, y_rk,v_rk = runge_kutta_4(m, c, k, f, y0, v0, t0, tf, h)
    plt.plot(t_rk, y_rk, label=f"h= {h}")
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Runge Kutta (RK4) Graph for Position')
y_eq = f / k #position equilibrium
plt.axhline(y_eq, linestyle='--', label="Position Equilibrium")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#RK4 Plotting (VELOCITY)
plt.figure()
for h in h_steps:
        t_rk, y_rk, v_rk = runge_kutta_4(m,c,k,f,y0,v0,t0,tf,h)
        plt.plot(t_rk, v_rk, label=f"h={h}")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Runge Kutta (RK4) Graph for Velocity")
v_eq = 0 
plt.axhline(v_eq, linestyle='--', label="Velocity Equilibrium (v=0)")
plt.grid(True)
plt.legend()
plt.show()
```

# Section 5: Stability Analysis

```python
#Demonstration of Stability Limits for Euler, demonstration is based off of growth vs decay relative to equilibrium, want less than a 10% deviation from the equilirbium at the end
y_eq = f / k
h_testarray = [2, 1, 0.5, 0.25, 0.1, 0.05, 0.01, 0.005]#testing h sizes for stability analysis
h_theory = 2 * m / c
print("Theoretical Euler stability scale (2m/c):", h_theory)
euler_growth = []
euler_max_amps = []
prev_h = None
stable_limit_found = False #used chat to create this condition because i was facing issues
for h in h_testarray:
    t, y, v = euler_forward(m, c, k, f, y0, v0, x_start, x_end, h)
    early_max = np.max(np.abs(y[:int(0.2 * len(y))]))
    late_max = np.max(np.abs(y[int(0.8 * len(y)):]))
    growth_ratio = late_max / early_max
    max_amp = np.max(np.abs(y))
    euler_growth.append(growth_ratio)
    euler_max_amps.append(max_amp)
    print(f"h = {h}")
    print(f"  growth ratio = {growth_ratio:.3f}")
    print(f"  max amplitude = {max_amp:.4f}")
    #theory check
    if h > 0.1 * h_theory:
        print("  theory prediction: UNSTABLE REGION (h too large relative to 2m/c)")
    elif h > 0.01 * h_theory:
        print("  theory prediction: borderline stability region")
    else:
        print("  theory prediction: stable based on damping scale")

    #empherical check
    if growth_ratio > 1.5:
        print("  behavior: UNSTABLE (non-physical growth)")
        if not stable_limit_found:
            print("Stability transition occurs between h ≈ 0.5 and h ≈ 1")
            stable_limit_found = True
    elif growth_ratio > 1.05:
        print("  behavior: marginally stable (numerical drift)")
    else:
        print("  behavior: stable (physically decaying)")
```

```python
#Demonstration of Stability limits for RK4
# RK4 Stability Analysis
rk4_growth = []
rk4_max_amps = []
for h in h_testarray:
    t, y, v = runge_kutta_4(m, c, k, f, y0, v0, x_start, x_end, h)
    early_max = np.max(np.abs(y[:int(0.2 * len(y))]))
    late_max = np.max(np.abs(y[int(0.8 * len(y)):]))
    growth_ratio = late_max / early_max
    max_amp = np.max(np.abs(y))
    rk4_growth.append(growth_ratio)
    rk4_max_amps.append(max_amp)
    print(f"h = {h}")
    print(f"  growth ratio = {growth_ratio:.3f}")
    print(f"  max amplitude = {max_amp:.4f}")
    if growth_ratio > 1.5:
        print("  behavior: UNSTABLE (non-physical growth)")
    elif growth_ratio > 1.05:
        print("  behavior: marginally stable")
    else:
        print("  behavior: stable (physically consistent)")
print("Euler becomes unstable for h > ~0.5–1, while RK4 remains stable for all tested step sizes.")

#theoretical stability:
print("\nTheoretical Stabiliy")
A = np.array([[0, 1],[-k/m, -c/m]])
eigvals = np.linalg.eigvals(A)
def R(z):
    return 1 + z + z**2/2 + z**3/6 + z**4/24
for h in h_testarray:
    print(f"\n h = {h}")
    for lam in eigvals:
        z = h * lam
        R_val = R(z)
        print("lambda:", lam)
        print("|R(z)|:", np.abs(R_val))
        if np.abs(R_val) <= 1:
            print("  theory: stable")
        else:
            print("  theory: unstable")
```

```python
#stability visualization:
plt.figure(figsize=(8,5))

plt.plot(h_testarray, euler_max_amps, 'o-', label="Euler max amplitude")
plt.plot(h_testarray, rk4_max_amps, 'o-', label="RK4 max amplitude")

plt.xscale("log")

plt.xlabel("Step size (h)")
plt.ylabel("Max amplitude")
plt.title("Growth-Based Stability Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#reference solution, extremely small steps to demonstrate accuracy and a similarity to the real solution
h_ref = 0.001
t_ref, y_ref, v_ref = runge_kutta_4(m, c, k, f, y0, v0, t0, tf, h_ref)
```

```python
#calculating error per each h step and then plotting it for euler and runge kutta
#rk4 error
for h in h_steps:
    # RK4 (choose one method per plot)
    t, y, v = runge_kutta_4(m, c, k, f, y0, v0, x_start, x_end, h)
    # interpolate onto reference grid
    y_interp = np.interp(t_ref, t, y)
    # pointwise error
    error = np.abs(y_interp - y_ref)
    plt.plot(t_ref, error, label=f"h = {h}")
plt.xlabel("Time (s)")
plt.ylabel("Absolute Error |y_h(t) - y_ref(t)|")
plt.title("Error Evolution Over Time for Different Step Sizes (RK4)")
plt.legend()
plt.grid(True)
plt.show()

#eulers error
for h in h_steps:
    t, y, v = euler_forward(m, c, k, f, y0, v0, x_start, x_end, h)
    # interpolate onto reference grid
    y_interp = np.interp(t_ref, t, y)
    # pointwise error
    error = np.abs(y_interp - y_ref)
    plt.plot(t_ref, error, label=f"h = {h}")
plt.xlabel("Time (s)")
plt.ylabel("Absolute Error |y_h(t) - y_ref(t)|")
plt.title("Error Evolution Over Time for Different Step Sizes (Eulers Forward)")
plt.legend()
plt.grid(True)
plt.show()
```

#Discussion of Stable vs Unstable

The stability is heavily dependent on the h step size as well as the method being used. As demonstrated in the stability calculations for both Euler and RK4, higher step sizes are more unstable with higher growth ratios and higher max amplitudes from the reference solution. Stability for Euler means that numerical errors are not growing uncontrollably and the solution remains physically consistent with the expected damped oscillatory behavior. This can be demonstrated in the high h values as they have increasing growth ratios (ratios above 1), suggesting that the solution is not correctly maintaining the expected decay toward equilibrium.For RK4, stability means that the numerical solution continues to produce bounded, physically realistic oscillations that match the expected damped behavior of the system. Since RK4 is a higher-order method with a larger stability region than Euler’s method, it can maintain accurate behavior for larger step sizes.

Regarding the error plots, this same concept can be demonstrated. Euler’s method shows larger oscillation error and reaches higher levels of absolute error due to its lower-order accuracy. On the other hand, RK4 shows a faster reduction in numerical error and better agreement with the reference solution. This demonstrates RK4’s improved stability and accuracy compared to Euler’s forward method. However, both methods show acceptable behavior for small step sizes such as h=0.01,0.05, and 0.1, since the numerical solutions remain close to the expected damped response, with Euler requiring smaller step sizes to achieve similar accuracy.

Regarding theoretical stability, the Euler method does not have a true stability threshold given by h=2m/c. Instead, 2m/c represents a physical time scale associated with damping in the system, not a numerical stability condition. Euler’s stability is actually governed by how the step size interacts with the eigenvalues of the system matrix, which determine whether numerical errors grow or decay.For RK4, stability is determined using the amplification function. The method is considered stable when: ∣R(z)∣≤1 for all eigenvalues λ of the system matrix. This condition aligns with the empirical results, where RK4 remains stable across all tested step sizes while Euler’s method becomes unstable for larger values of h.
