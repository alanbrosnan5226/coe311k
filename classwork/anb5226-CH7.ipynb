```python
import numpy as np
def jacobi(A, b, x0, tol, max_iterations=100000):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
def gauss_seidel(A, b, x0, tol, max_iterations=100000):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
# Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k 
        x = x_new
    return x, max_iterations
```

```python
#SET A
A= np.array([[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)
b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)
x0=np.zeros_like(b,dtype=float)
tol=1e-6

solution, iterations= jacobi(A,b,x0,tol)
print ("Set A Solution Jacobi:", solution,"\n","Set A Iterations Jacobi:", iterations)
solution2, iterations2= gauss_seidel(A,b,x0,tol)
print ("Set A Solution Gauss_Seidel:", solution2,"\n","Set A Iterations Gauss_Seidel:", iterations2)
```

```python
#Set B
C = np.array([[1, 1], [1, 1.0001]])
d = np.array([1, 1])
x0=np.zeros_like(d,dtype=float)
solution3, iterations3= jacobi(C,d,x0,tol)
print ("Set B Solution Jacobi:", solution3,"\n","Set B Iterations Jacobi:", iterations3)
solution4, iterations4= gauss_seidel(C,d,x0,tol)
print ("Set B Solution Gauss_Seidel:", solution4,"\n","Set B Iterations Gauss_Seidel:", iterations4)
```

```python
#Set C
E= np.array([[1, 1], [1, 1.0001]])
f= np.array([1, 1.0001])
x0=np.zeros_like(f,dtype=float)
solution5, iterations5= jacobi (E,f,x0,tol)
print ("Set C Solution Jacobi:", solution5,"\n","Set C Iterations Jacobi:", iterations5)
solution6, iterations6= gauss_seidel(E,f,x0,tol)
print ("Set C Solution Gauss_Seidel", solution6,"\n","Set C Iterations Gauss_Seidel:", iterations6)
```
