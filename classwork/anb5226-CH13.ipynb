```python
#Use custom matrix multiplier form lecture and time each pair
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
#optimized method (dot product)
#result_optimized= np.dot(array1,array2)
```

```python
#Data Collection
sizes=[]
naive_times=[]
optimized_times=[]
n=3

for i in range (8):
    A,B= building_matrix(n)
    #Naive Time
    start_naive= time.perf_counter()
    naive=naive_matrix_multiplication(A,B)
    end_naive= time.perf_counter()
    naive_times.append(end_naive-start_naive)
    
    start_optimized= time.perf_counter()
    optimized=np.dot(A,B)
    end_optimized= time.perf_counter()
    optimized_times.append(end_optimized-start_optimized)
    
    print ("Pairs of size: ", n,"*",n, "\n Matrix 1: \n", A, "\n Matrix 2: \n", B, "\n")
    sizes.append(n)
    n=n*2
```

```python
#Compare time to numpy and Plot
n = np.linspace(1, 100, 100)
plt.plot(sizes, naive_times, label='Naive', marker='o')
plt.plot(sizes, optimized_times, label='Optimized', marker='s')
plt.yscale('log') # Logarithmic scale for better visualization
plt.title('Naive vs Optimized Run Time')
plt.xlabel('Matrix Sizes')
plt.ylabel('Time')
plt.legend()
plt.show()
```
