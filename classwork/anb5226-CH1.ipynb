```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.linspace(0, 2*np.pi, 50)
y= np.sin(x)
z= np.cos(x)
plt.subplot (2,2,1)
plt.plot(x, np.sin(x),'r')
plt.subplot (2,2,2)
plt.plot(x, np.cos(x),'r')
plt.subplot(2,2,3)
plt.scatter(x,y*y,)
plt.subplot(2,2,4)
plt.scatter(x,z*z,)
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
x= np.linspace(0, 2, 20)
y= (x**2)
plt.subplot(2,2,1)
plt.hist(y,50)
plt.subplot (2,2,2)
plt.scatter (x,y)
plt.subplot (2,2,3)
plt.plot (x,y)
plt.subplot (2,2,4)
plt.plot(x**x,y**y)

```
