```python
import random
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def picalc (numsamples):
    counter= 0
    
    for i in range (0, numsamples):
        x= random.uniform (-1, 1)
        y= random.uniform (-1, 1)
        
        if (x*x + y*y <1):
            counter += 1
    
    pi_est= 4* (counter/numsamples)
    
    return pi_est
        
```

```python
true_error= []
relative_error= []
true_pi= np.pi

for i in range (0,100):
    calculated_pi= picalc (1000)
    t_error= abs(true_pi-calculated_pi)
    r_error= abs ((true_pi-calculated_pi)/true_pi)
    
    true_error.append (t_error)
    relative_error.append(r_error)
```

```python
plt.subplot (3,1,1)
plt.plot (true_error)
plt.xlabel ("Sample Number")
plt.ylabel ("True Error")
plt.title ("True Error of Calculation")



plt.subplot (3,1,2)
plt.plot (relative_error)
plt.xlabel ("Sample Number")
plt.ylabel ("Relative Error")
plt.title ("Relative Error of Calculation")



plt.subplot (3,1,3)
plt.plot (true_error)
plt.plot (relative_error)
plt.xlabel ("True Error")
plt.ylabel ("Relative Error")
plt.title ("True and Relative Error")

plt.tight_layout()
plt.show()
```
