```python
import numpy as np
import matplotlib.pyplot as plt
#approximative derivative of sin(x)
#include 4 plots, cos, forward, backward, and central

#Constants(A,B, and Samples):
f = lambda x: np.sin(x)
a=0 #start
b=2*np.pi #endpoint
N=50 #samples ize

#uniform grid
x = np.linspace(a, b, N)
y = f(x)

# Step size
h = (b - a) / (N - 1)
# Central Difference Approximation for the derivative
df_dx = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
df_dxf = (f(x+h)-f(x))/h
# Backward Difference Approximation
df_dxb = (f(x)-f(x-h))/h
```

```python
# Plot the approximation and the true derivative
plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx, 'o--', label='Central Difference Approximation')
plt.plot(x, df_dxf, 'b--', label='Forward Difference Approximation')
plt.plot(x, df_dxb, 'r--', label= 'Backward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Central Difference")
plt.legend()
plt.show()
```
