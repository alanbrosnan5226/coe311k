Jacobi

```python
import numpy as np
import time
```

```python
#Matrixes to work with
A= np.array([[10,-1,2,0,0],
             [-1,11,-1,3,0],
             [2,-1,10,-1,0],
             [0,3,-1,8,-2],
             [0,0,0,-2,9]], dtype=float)
b= np.array([14,30,26,25,37], dtype= float)
```

```python
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
x0= np.zeros_like(b)
tolerance = 1e-6
max_iterations= 100
start= time.time()
solution, iterations= jacobi(A, b, x0, tolerance, max_iterations)
end= time.time()
print (f"Jacobi Solution: ",solution)
print (f"Jacobi Iterations: ", iterations)
print (f"Jacobi Time:", (end-start))
```

Gauss-Seidel

```python
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
# Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
x02= np.zeros_like(b)
tolerance2= 1e-6
max_iterations2= 100
start2=time.time()
solution,iterations= gauss_seidel(A,b,x02,tolerance2,max_iterations)
end2=time.time()
print(f"Solution to Gauss-Seidel Equation: ", solution)
print(f"Number of iterations for Gauss-Seidel: ", iterations)
print(f"Time for Gaussian-Seidel: ", end2-start2)
```

For the Jacobi function it took 18 iterations to achieve the answers x is approximately equal to [1,2,3,4,5].It took approximately 0.0011 seconds to run the function. For the gauss_seidel function it took less iterations, only taking 10 iterations to achieve the same answer. It only took 0.0008 seconds to achieve the same answer, showing that the Gauss-Seidel method is faster.
