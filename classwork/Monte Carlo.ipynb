Monte Carlo Method Jan. 29 2026

4*(Number of Hits/ Number of Points)

```python
def MonteCarloApprox (numSamples):
    import random
    import math
    
    counter=0
    
    for i in range (0, numSamples):
        x= random.uniform (-1,1)
        y= random.uniform (-1,1)
    
        if (x*x + y*y <1):
            counter += 1
            
    return 4 * (counter/numSamples)
```

```python
numSamples = 10000000

MonteCarloApprox(numSamples)
```
