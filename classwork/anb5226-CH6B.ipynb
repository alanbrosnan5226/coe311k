```python
import numpy as np
import time
```

```python
A= np.array([[1,2,3,0,0],[2,1,2,3,0],[3,2,1,2,3],[0,3,2,1,2],[0,0,3,2,1]], dtype=float)
b= np.array([14,22,33,26,22],dtype=float)
```

```python
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
x0= np.zeros_like(b)
tolerance = 1e-6
max_iterations= 100
start= time.time()
solution, iterations= jacobi(A, b, x0, tolerance, max_iterations)
end= time.time()
print ("Jacobi Solution: ",solution)
print ("Jacobi Iterations: ", iterations)
print ("Jacobi Time:", (end-start))
```

```python
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
# Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
x02= np.zeros_like(b)
tolerance2= 1e-6
max_iterations2= 100
start2=time.time()
solution,iterations= gauss_seidel(A,b,x02,tolerance2,max_iterations)
end2=time.time()
print("Solution to Gauss-Seidel Equation: ", solution)
print("Number of iterations for Gauss-Seidel: ", iterations)
print("Time for Gaussian-Seidel: ", end2-start2)
```

Both of the functions found that the solutions diverged. Both of the functions reached the maximum iterations, meaning that the matrix diverges. The jacobi matrix function found that it took 0.0079 seconds to reach the maximum, while the gauss_seidel function found it took 0.0046 seconds to reach the max iterations.
